# 图的介绍

## 🔗 NetworkX 中的无向图（Undirected Graph）

无向图（Undirected Graph）是一种**边没有方向性**的图结构，表示两个节点之间的关系是**对等的、双向的**。

### ✅ 与有向图的区别：

| 图类型       | 描述                                       | 示例         |
|--------------|--------------------------------------------|--------------|
| 有向图 DiGraph | 边有方向，A → B 与 B → A 是两条不同的边    | 微博关注关系 |
| 无向图 Graph   | 边无方向，A—B 与 B—A 是同一条边            | 社交好友关系、共现词图 |




### 📌 NetworkX 中如何创建无向图：

```python
import networkx as nx

G = nx.Graph()  # 无向图
G.add_nodes_from([1, 2, 3])
G.add_edges_from([(1, 2), (2, 3)])


````python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

 
# 创建一个空的无向图
G = nx.Graph()
 
# 添加节点
G.add_nodes_from([1, 2, 3, 4])
 
# 添加边
G.add_edges_from([(1, 2), (2, 3), (3, 4), (4, 1)])
 
# 或者可以使用下面的方式一次性添加边
# edges = [(1, 2), (2, 3), (3, 4), (4, 1)]
# G.add_edges_from(edges)
nx.draw(G, with_labels=True)
plt.show()
````

````python
import networkx as nx
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体显示中文
plt.rcParams['axes.unicode_minus'] = False    # 正常显示负号

# 创建无向图对象
G = nx.Graph()

# 添加节点
G.add_nodes_from(["A", "B", "C", "D"])

# 添加无向边
G.add_edges_from([("A", "B"), ("B", "C"), ("C", "D"), ("D", "A")])

# 可视化布局
pos = nx.spring_layout(G, seed=42)

# 绘图
plt.figure(figsize=(5, 4))
nx.draw(G, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=800)
plt.title("NetworkX 无向图示意")
plt.show()
````

## ➡️ NetworkX 中的有向图（Directed Graph）

有向图（Directed Graph）是一种**边具有方向性**的图结构，表示节点之间的关系是**单向的**，即 A → B 不等于 B → A。


### ✅ 与无向图的区别：

| 图类型       | 描述                                    | 举例                 |
|--------------|-----------------------------------------|----------------------|
| 有向图 DiGraph | A → B 表示 A 指向 B，不代表 B 指向 A     | 微博关注、依赖关系   |
| 无向图 Graph   | A—B 与 B—A 等价，表示双向关系           | 好友关系、合作网络   |





### 📌 如何创建有向图（使用 `DiGraph`）：

```python
import networkx as nx

G = nx.DiGraph()  # 创建有向图对象
G.add_edges_from([("A", "B"), ("B", "C")])


````python
import networkx as nx
 
# 创建一个空的有向图
G = nx.DiGraph()
 
# 添加节点
G.add_nodes_from([1, 2, 3, 4])
 
# 添加有向边
G.add_edges_from([(1, 2), (2, 3), (3, 4), (4, 1)])
 
# 或者可以使用下面的方式一次性添加有向边
# edges = [(1, 2), (2, 3), (3, 4), (4, 1)]
# G.add_edges_from(edges)
nx.draw(G, with_labels=True)
plt.show() 
````

````python
import networkx as nx
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体显示中文
plt.rcParams['axes.unicode_minus'] = False    # 正常显示负号
import warnings
warnings.filterwarnings('ignore')

# 创建有向图对象
G = nx.DiGraph()

# 添加节点与有向边
G.add_edges_from([
    ("任务A", "任务B"),
    ("任务B", "任务C"),
    ("任务A", "任务C"),
    ("任务C", "任务D")
])

# 生成节点布局
pos = nx.spring_layout(G, seed=42)

# 绘制有向图
plt.figure(figsize=(6, 4))
nx.draw(G, pos, with_labels=True, node_color='lightgreen', edge_color='black',
        node_size=800, arrows=True, arrowsize=20)
plt.title("有向图示意（任务依赖）")
plt.show()
````

## ⚖️ NetworkX 中的加权图（Weighted Graph）

加权图是指**每条边都附带一个权重（weight）**的图，权重可以表示强度、频率、距离、成本等含义。

### ✅ 加权图支持两种结构：
- 无向加权图（Graph）
- 有向加权图（DiGraph）



### 📌 应用场景举例：

| 应用领域         | 边的权重含义             |
|------------------|--------------------------|
| 社交网络         | 两人之间互动频次         |
| 公路网络         | 城市间距离或通行时间     |
| 情绪传播图       | 情感共现频率             |
| 知识图谱         | 关系强度或置信度         |

---



### 📎 如何创建加权图（示例为无向图）：

```python
import networkx as nx

G = nx.Graph()  # 创建无向图

# 添加带权边：节点1, 节点2, 权重
G.add_weighted_edges_from([
    ("A", "B", 1.5),
    ("B", "C", 2.0),
    ("C", "D", 0.8),
    ("A", "D", 1.2)
])


````python
import networkx as nx
 
# 创建一个空的加权图
G = nx.Graph()
 
# 添加节点
G.add_nodes_from([1, 2, 3, 4])
 
# 添加带权边
G.add_weighted_edges_from([(1, 2, 0.5), (2, 3, 1.2), (3, 4, 0.8), (4, 1, 1.5)])
 
# 或者可以使用下面的方式一次性添加带权边
# weighted_edges = [(1, 2, 0.5), (2, 3, 1.2), (3, 4, 0.8), (4, 1, 1.5)]
# G.add_weighted_edges_from(weighted_edges)
nx.draw(G, with_labels=True)
plt.show() 
````

````python
import networkx as nx
import matplotlib.pyplot as plt



G = nx.Graph()

# 添加带权边：(起点, 终点, 权重值)
# NetworkX 自动把权重值存进字典，键名默认叫 'weight'
edges = [(1, 2, 0.5), (2, 3, 1.2), (3, 4, 0.8), (4, 1, 1.5)]
G.add_weighted_edges_from(edges)

# --------------------------------------------------
# 🔍 关键演示：看看每条边背后藏着什么？
# --------------------------------------------------
print("=" * 50)
print("不带 data=True，只能看到端点：")
for item in G.edges():
    print(f"  {item}")

print("\n" + "=" * 50)
print("带上 data=True，能看到隐藏的字典：")
for u, v, d in G.edges(data=True):
    print(f"  边 ({u}→{v})，属性字典 d = {d}")

print("\n" + "=" * 50)
print("用 d['weight'] 取出权重值：")
for u, v, d in G.edges(data=True):
    print(f"  边 ({u}→{v}) 的权重 = {d['weight']}")

print("\n" + "=" * 50)
print("权重 × 2 得到画图用的线宽：")
for u, v, d in G.edges(data=True):
    w = d['weight']
    print(f"  边 ({u}→{v}): 权重={w}, 线宽={w * 2}")
````

````python
import networkx as nx
import matplotlib.pyplot as plt

# 1. 创建无向加权图
G = nx.Graph()
G.add_weighted_edges_from([
    ("北京", "上海", 1200),
    ("上海", "广州", 900),
    ("广州","深圳", 150),
    ("北京", "广州", 2000)
])

# --------------------------------------------------
# 🔑 核心步骤：提前把权重提取出来，做成一个字典
# weights 的长相：{('北京', '上海'): 1200, ('上海', '广州'): 900, ...}
# --------------------------------------------------
weights = nx.get_edge_attributes(G, "weight")

# 复用 1：通过查字典，算出每条边的线宽
# G.edges() 只给边名，如 ('北京', '上海')，拿它去 weights 字典里查对应的值
edge_widths = [weights[edge] / 400 for edge in G.edges()]

# 2. 生成节点布局
pos = nx.spring_layout(G, seed=42)

# 3. 绘图
plt.figure(figsize=(8, 6))

# 画节点和边（线宽用刚才算的 edge_widths）
nx.draw(G, pos, with_labels=True, node_color="lightyellow", edge_color="gray",
        width=edge_widths, node_size=800, font_family='SimHei') # SimHei防中文乱码

# 复用 2：画边上的标签，直接把 weights 字典扔给 edge_labels
# 因为 edge_labels 需要的格式正是 {边: 文字}，而 weights 天生就是这个格式！
nx.draw_networkx_edge_labels(G, pos, edge_labels=weights, font_size=10)

plt.title("方法二：提取字典复用（线宽+标签共用权重）")
plt.show()
````


#### 提取边权重的两种方式：该用哪种？
在 NetworkX 中，如果我们想根据权重来调整画图效果（比如线变粗、显示数字），我们需要把权重从图里“拿”出来。有两种拿法：


##### 🥷 方式一：边走边拿（游击战）

````python
edge_width = [d['weight'] * 2 for (u, v, d) in G.edges(data=True)]
print(edge_width)
````

**工作原理：**
- `data=True` 让我们背上了一个叫 `d` 的属性背包。
- 遍历每条边时，当场拉开背包 `d['weight']` 拿出权重，算成线宽。
- 算完之后，背包就扔了，下次想用还得重新遍历。
**特点：** 代码短，一步到位。**适合：只用一次权重的场景。**

##### 🏪 方式二：先建仓库再查字典（正规军）

````python
# 第一步：建仓库（提取成字典）
weights = nx.get_edge_attributes(G, "weight")
print(weights)
# 第二步：按需查字典
edge_widths = [weights[edge] / 400 for edge in G.edges()]  # 用一次
print(edge_widths)
nx.draw_networkx_edge_labels(G, pos, edge_labels=weights)   # 再用一次！
````


**工作原理：**
- 先把所有权重单独抽出来，做成一个专门的字典 `weights`。
- 字典长这样：`{('北京', '上海'): 1200, ('上海', '广州'): 900}`
- 以后不管哪里要用权重，只要拿着边名（如 `('北京', '上海')`）去字典里查就行。
**特点：** 数据可以复用，代码语义清晰。**适合：多次使用权重的场景。**

##### 🤔 核心区别：画标签的硬性要求
为什么方式二在画图时更常用？因为**画边标签 `edge_labels` 强迫你必须传字典！**
- 画标签函数需要知道“**写在哪条边上**”和“**写什么数字**”。
- 列表只有数字，没有边名，函数会懵：`[1200, 900, 150, 2000]` —— 这 1200 是给谁看的？
- 字典有对应关系，函数能看懂：`{('北京', '上海'): 1200}` —— 哦！1200 写在这条边上！



##### 📋 总结口诀
| 场景 | 推荐方式 | 理由 |
| :--- | :--- | :--- |
| 只改线宽/颜色，不画标签 | 方式一 `(u,v,d)` | 代码简短，省事 |
| **又要改线宽，又要画标签** | **方式二 `get_edge_attributes`** | **提取一次，到处复用，且画标签必须传字典** |

````python
import networkx as nx
import matplotlib.pyplot as plt

# 创建一个空的加权图
G = nx.Graph()

# 添加节点
G.add_nodes_from([1, 2, 3, 4])

# 添加带权边
edges = [(1, 2, 0.5), (2, 3, 1.2), (3, 4, 0.8), (4, 1, 1.5)]
G.add_weighted_edges_from(edges)

# 绘制图形
pos = nx.spring_layout(G)  # 使用 spring 布局

# 根据权重设置边的宽度
edge_width = [d['weight'] * 2 for (u, v, d) in G.edges(data=True)]

# 根据权重设置边的颜色
edge_color = [d['weight'] for (u, v, d) in G.edges(data=True)]

# 绘制节点
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=500)

# 绘制边
nx.draw_networkx_edges(G, pos, width=edge_width, edge_color=edge_color, edge_cmap=plt.cm.Blues)

# 绘制标签
nx.draw_networkx_labels(G, pos, font_size=10, font_color='black')

# 添加边的权重标签
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

# 显示图形
plt.show()
````

````python
import networkx as nx
import matplotlib.pyplot as plt

# 创建无向加权图
G = nx.Graph()
G.add_weighted_edges_from([
    ("A", "B", 1.5),
    ("B", "C", 2.0),
    ("C", "D", 0.8),
    ("A", "D", 1.2)
])

# 提取边权重作为线宽
weights = nx.get_edge_attributes(G, "weight")
edge_widths = [weights[edge] * 2.5 for edge in G.edges]

# 生成节点布局
pos = nx.spring_layout(G, seed=42)

# 绘图：边宽体现权重
plt.figure(figsize=(6, 5))
nx.draw(G, pos, with_labels=True, node_color="skyblue", edge_color="gray",
        width=edge_widths, node_size=800)
nx.draw_networkx_edge_labels(G, pos, edge_labels=weights)
plt.title("无向加权图：边宽代表权重")
plt.show()
````

## ♻️ 多重图 Multigraph（MultiGraph / MultiDiGraph）

多重图是一种**允许两个节点之间存在多条边**的图结构，边可以有不同的属性、权重、类型等。

---

### ✅ 与普通图的区别：

| 类型          | 是否支持重复边 | 是否支持方向 | 使用类名         |
|---------------|----------------|---------------|------------------|
| Graph         | ❌ 否           | ❌ 无方向      | `nx.Graph()`     |
| DiGraph       | ❌ 否           | ✅ 有方向      | `nx.DiGraph()`   |
| MultiGraph    | ✅ 是           | ❌ 无方向      | `nx.MultiGraph()`|
| MultiDiGraph  | ✅ 是           | ✅ 有方向      | `nx.MultiDiGraph()`|

---

### 📌 应用场景：

- 同一对节点之间存在多种“关系”：如“合作”+“引用”
- 表示交通网络中不同时段的连接
- 社交网络中“多轮互动”、“多次交易”
- 多种共现路径（如：词语A–B 既同句共现也在标题共现）

---

### 🧠 多重图的边是有编号的（edge keys）：

每条边都有唯一的 `(u, v, key)` 三元组标识，其中 `key` 表示第几条边。


````python
import networkx as nx
import matplotlib.pyplot as plt

# 创建一个空的多重无向图
G = nx.MultiGraph()

# 添加节点
G.add_nodes_from([1, 2, 3, 4])

# 添加多重边
G.add_edges_from([(1, 2), (1, 2), (2, 3), (3, 4)])

# 获取节点位置
pos = nx.spring_layout(G)

# 绘制节点
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=500)

# 手动绘制多重边
# 使用不同的弧度来区分多重边
edges = list(G.edges(data=True))
for i, (u, v, data) in enumerate(edges):
    rad = 0.2 * (i + 1)  # 调整弧度以区分多重边
    nx.draw_networkx_edges(G, pos, edgelist=[(u, v)], connectionstyle=f'arc3,rad={rad}')

# 绘制标签
nx.draw_networkx_labels(G, pos, font_size=10, font_color='black')

# 显示图形
plt.show()
````

````python
import networkx as nx
import matplotlib.pyplot as plt

# 创建无向多重图
G = nx.MultiGraph()

# 添加节点
G.add_nodes_from(["A", "B"])

# 添加多条边（同一对节点）
G.add_edge("A", "B", weight=1.0, relation="合作")
G.add_edge("A", "B", weight=2.0, relation="引用")
G.add_edge("A", "B", weight=0.8, relation="朋友")

# 生成布局
pos = nx.spring_layout(G, seed=42)

# 绘图设置
plt.figure(figsize=(6, 5))
ax = plt.gca()

# 绘制节点
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=800, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=12)

# 绘制多条边，用不同弧度区分
edges = list(G.edges(keys=True, data=True))
for i, (u, v, k, data) in enumerate(edges):
    rad = 0.2 * (i - 1)  # 控制每条边的弯曲程度
    nx.draw_networkx_edges(G, pos, edgelist=[(u, v)], connectionstyle=f'arc3,rad={rad}', ax=ax)

    # 添加每条边的文本标签（权重 + 关系）
    mid_x = (pos[u][0] + pos[v][0]) / 2
    mid_y = (pos[u][1] + pos[v][1]) / 2 + rad
    label = f"{data['relation']} ({data['weight']})"
    plt.text(mid_x, mid_y, label, fontsize=9, color='black', ha='center', va='center')

plt.title("多重图示例：A-B 多种关系")
plt.axis("off")
plt.tight_layout()
plt.show()
````

###  `arc3` 是什么意思？
`arc3` 是 Matplotlib 画图库里的一个术语，用来控制**连线的弯曲方式**。
* **`arc`**：弧线。
* **`3`**：代表“三次贝塞尔曲线”（3rd order Bezier curve）。你不用管贝塞尔曲线多复杂，只要知道：**它能让线条画出平滑的弧度**。
后面的 `rad=...` 是 radius（半径/弧度）的缩写，控制**弯的程度和方向**：
* `rad=0`：直线（不弯）
* `rad=正数`：向上弯（凸起）
* `rad=负数`：向下弯（凹陷）
**为什么这里要用 `arc3`？**
因为这是**多重图**，A 和 B 之间有 3 条边。如果都是直线，这 3 条边会**完全重叠**在一起，你只能看到一条线。
代码里用了 `rad = 0.2 * (i - 1)`，这招很巧妙：
* 第 1 条边 (`i=0`)：`rad = -0.2`，向下弯
* 第 2 条边 (`i=1`)：`rad = 0`，直线
* 第 3 条边 (`i=2`)：`rad = 0.2`，向上弯
这样 3 条边就错开啦！



---
### `(pos[u][0] + pos[v][0])` 是什么意思？
这是一道简单的**数学求中点**题。它的目的是：**算出这条边的中间位置，好把文字标签贴上去。**
我们一层层剥开看：
#### `pos` 是什么？
`pos` 是 `nx.spring_layout(G)` 算出来的**节点坐标字典**。它长这样：
```python
{
    'A': [0.1, 0.8],  # A 的 X 坐标是 0.1，Y 坐标是 0.8
    'B': [0.9, 0.2]   # B 的 X 坐标是 0.9，Y 坐标是 0.2
}
```
#### `pos[u]` 和 `pos[v]` 是什么？
在循环里，`u` 和 `v` 是边的两端节点（比如 `u='A'`, `v='B'`）。
* `pos[u]` 就是 `pos['A']`，也就是 `[0.1, 0.8]`
* `pos[v]` 就是 `pos['B']`，也就是 `[0.9, 0.2]`
#### `[0]` 是什么意思？
列表可以通过索引取值，`[0]` 取的是第一个元素，也就是 **X 坐标**！
* `pos['A'][0]` = 0.1 （A 的 X 坐标）
* `pos['B'][0]` = 0.9 （B 的 X 坐标）
#### 合起来看
```python
mid_x = (pos[u][0] + pos[v][0]) / 2
# 代入数字：(0.1 + 0.9) / 2 = 1.0 / 2 = 0.5
# 这就是 A 和 B 中点位置的 X 坐标！
```
同理，下一行的：
```python
mid_y = (pos[u][1] + pos[v][1]) / 2
# [1] 取的是 Y 坐标
# 代入数字：(0.8 + 0.2) / 2 = 0.5
# 这就是 A 和 B 中点位置的 Y 坐标！
```
#### 算出 `mid_x = 0.5` 和 `mid_y = 0.5` 后，代码接着用 `plt.text(mid_x, mid_y, ...)`，就把文字精准地贴在了 A 和 B 连线的中点位置。后面加的 `+ rad`，是为了让文字跟着弧线一起偏移一点，避免文字和线重叠。

## 🔁 带环图（Cyclic Graph）

带环图是指**图中至少存在一条闭合路径（回路）**，即：
从某个节点出发，经过若干条边，最终又回到起点。

---

### ✅ 特征：

- 至少存在一个 **简单环（cycle）**
- 可以是无向图或有向图（分别称为“无向带环图”和“有向带环图”）
- 与“无环图（Acyclic Graph）”相对（如 DAG）

---

### 📎 应用举例：

| 应用场景         | 含义或用途                          |
|------------------|-------------------------------------|
| 项目流程图       | 若出现环则说明存在“循环依赖”        |
| 地铁交通网络     | 某些站点之间存在环形线路            |
| 社交推荐系统     | A→B→C→A 可用于检测“互相关注圈”     |

---

NetworkX 提供了多种布局算法，用于绘制网络图。以下是一些常见的布局方式：

- Spring Layout (spring_layout)：基于 Fruchterman-Reingold 算法，模拟弹簧系统。
- Circular Layout (circular_layout)：将节点排列在一个圆周上。
- Spectral Layout (spectral_layout)：基于图的拉普拉斯矩阵的特征向量。
- Random Layout (random_layout)：节点随机排列。
- Planar Layout (planar_layout)：适用于平面图的布局。

````python
import networkx as nx
import matplotlib.pyplot as plt

# 创建一个图 G
G = nx.Graph()

# 添加节点和边
G.add_edges_from([(0, 1), (0, 2), (0, 3), (1, 4), (1, 5), 
                  (2, 6), (2, 7), (3, 8), (3, 9), (4, 10), 
                  (5, 11), (6, 12), (7, 13), (8, 14), 
                  (9, 15)])

# 定义布局
layouts = {
    "Spring Layout": nx.spring_layout(G),
    "Circular Layout": nx.circular_layout(G),
    "Spectral Layout": nx.spectral_layout(G),
    "Random Layout": nx.random_layout(G),
    "Planar Layout": nx.planar_layout(G)
}

# 绘制图形
plt.figure(figsize=(15, 10))

for i, (layout_name, layout) in enumerate(layouts.items()):
    plt.subplot(2, 3, i + 1)
    nx.draw(G, pos=layout, with_labels=True, node_color='lightblue', node_size=500, edge_color='gray')
    plt.title(layout_name)

plt.subplot(2, 3, 6).axis('off')  # 去掉最后一个空位的坐标轴
plt.show()
````

## 数据质量检查

正式构建网络之前，应检查节点表和边表是否满足基本要求：字段是否完整，边表中的 source 和 target 是否都能在节点表中找到，是否存在缺失值等。


````python
# 必要字段检查
required_node_cols = {"node", "category", "importance"}
required_edge_cols = {"source", "target", "weight", "type"}

print("节点表字段是否完整：", required_node_cols.issubset(nodes_df.columns))
print("边表字段是否完整：", required_edge_cols.issubset(edges_df.columns))

# 检查边表中的节点是否都出现在节点表中
node_set = set(nodes_df["node"])
edge_node_set = set(edges_df["source"]).union(set(edges_df["target"]))
missing_nodes = edge_node_set - node_set

print("边表中未出现在节点表的节点：", missing_nodes)

# 如果有缺失节点，可选择补充到节点表，或检查边表是否写错
if missing_nodes:
    extra_nodes = pd.DataFrame({
        "node": list(missing_nodes),
        "category": ["待补充"] * len(missing_nodes),
        "importance": [1] * len(missing_nodes),
        "description": ["由边表自动识别的节点"] * len(missing_nodes)
    })
    nodes_df = pd.concat([nodes_df, extra_nodes], ignore_index=True)
    print("已自动补充缺失节点。")
````

## 根据主题选择图类型

不同主题应选择不同的图结构：

| 图类型 | NetworkX 对象 | 适用情形 |
|---|---|---|
| 无向图 | `nx.Graph()` | 同学关系、共同出现、合作关系等不强调方向的关系 |
| 有向图 | `nx.DiGraph()` | 供应链上下游、知识点依赖、信息传播、任务流程等强调方向的关系 |
| 多重无向图 | `nx.MultiGraph()` | 两个节点之间可能存在多种无向关系 |
| 多重有向图 | `nx.MultiDiGraph()` | 两个节点之间可能存在多种有方向关系 |

本例中的边包含“流程依赖”“教学关系”“供应链关系”“外部影响”等方向性关系，因此选择 `DiGraph()`。

````python
# 使用边表构建有向图
G = nx.from_pandas_edgelist(
    edges_df,
    source="source",
    target="target",
    edge_attr=["weight", "type", "note"],
    create_using=nx.DiGraph()
)

# 将节点表中的属性写入网络
for _, row in nodes_df.iterrows():
    G.add_node(
        row["node"],
        category=row["category"],
        importance=row["importance"],
        description=row.get("description", "")
    )

print("图类型：", type(G))
print("节点数：", G.number_of_nodes())
print("边数：", G.number_of_edges())

# 查看一个节点和一条边的属性
example_node = list(G.nodes())[0]
example_edge = list(G.edges(data=True))[0]
print("示例节点属性：", example_node, G.nodes[example_node])
print("示例边属性：", example_edge)
````

## 计算节点层面的网络指标

常用节点指标包括：

度数：一个节点连接了多少条边；  
入度 / 出度：在有向图中，分别表示接收关系和发出关系的数量；  
加权度：考虑边权重后的连接强度；  
度中心性：节点在整体网络中的直接连接能力；  
介数中心性：节点是否位于其他节点之间的关键路径上；  
接近中心性：节点到其他节点的平均距离是否较短。

````python
# 计算节点指标
degree_dict = dict(G.degree())
in_degree_dict = dict(G.in_degree())
out_degree_dict = dict(G.out_degree())
weighted_degree_dict = dict(G.degree(weight="weight"))

degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G, weight="weight", normalized=True)
closeness_centrality = nx.closeness_centrality(G)

metrics_df = pd.DataFrame({
    "node": list(G.nodes()),
    "category": [G.nodes[n].get("category") for n in G.nodes()],
    "degree": [degree_dict[n] for n in G.nodes()],
    "in_degree": [in_degree_dict[n] for n in G.nodes()],
    "out_degree": [out_degree_dict[n] for n in G.nodes()],
    "weighted_degree": [weighted_degree_dict[n] for n in G.nodes()],
    "degree_centrality": [degree_centrality[n] for n in G.nodes()],
    "betweenness_centrality": [betweenness_centrality[n] for n in G.nodes()],
    "closeness_centrality": [closeness_centrality[n] for n in G.nodes()]
})

metrics_df_sorted = metrics_df.sort_values("betweenness_centrality", ascending=False)
display(metrics_df_sorted.head(10))

# 保存指标结果
metrics_df_sorted.to_csv("网络指标结果.csv", index=False, encoding="utf-8-sig")
````

## 计算网络整体结构指标

整体指标可以帮助我们判断网络是否紧密、是否分散、是否存在多个子群体。常用指标包括：

网络密度：实际边数占可能边数的比例；  
平均聚类系数：节点邻居之间是否也相互连接；  
连通分量：网络是否分裂为多个部分；  
平均最短路径长度：节点之间平均需要经过多少步才能到达，通常只在连通子图上计算。

“强”和“弱”连通分量，**首先要强调一个大前提：这两个概念只存在于“有向图”中！**（无向图只有普通的“连通分量”）。

---
### 🚗 核心比喻：单行道 vs 双行道
想象图中的节点是城市，边是道路：
* **弱连通**：相当于**无视交通规则**（把所有单行道当成双行道），只要路连着，能走过去就算连通。
* **强连通**：相当于**严格遵守交通规则**（必须顺着箭头方向走），你从 A 能走到 B，**而且**还能顺着箭头从 B 走回 A，这才算强连通。
---
#### 弱连通分量
**定义：** 如果忽略图中所有边的方向，把图当成无向图来看，节点之间互相可达的最大子图。
**通俗解释：** “物理上连在一起”。
哪怕 A 到 B 只有单行道，回不来，但在弱连通看来，你们俩挨着，就算一个团伙。
#### 强连通分量
**定义：** 在有向图中，对于子图里的任意两个节点 u 和 v，u 能顺着箭头走到 v，且 v 也能顺着箭头走回 u。
**通俗解释：** “逻辑上能互相串门”。
不仅要去得了，还得回得来！**强连通分量里的任意两个节点，都必须是“双向奔赴”的。**
---
### 👀 看图说话（最直观的例子）
我们画一个极其简单的有向图：
```text
  1 ──→ 2 ⟶ 3
        ↑    ↓
        └────┘
```
* 1 指向 2
* 2 指向 3
* 3 指向 2
**分析：**
1. **弱连通分量：** `[1, 2, 3]`
   * 为什么？因为如果不看箭头，1、2、3 物理上是连在一起的，属于同一个弱连通分量。
2. **强连通分量：** `[2, 3]` 和 `[1]`
   * 为什么是 `[2, 3]`？因为从 2 能到 3，从 3 也能回到 2，它们完美闭环！
   * 为什么 1 不是？因为 1 能去 2，但**从 2 没有任何箭头能回到 1**！1 是个“光出不进”的孤岛，所以它自己单独算一个强连通分量。

---



````python
import networkx as nx
# 创建有向图
G = nx.DiGraph()
G.add_edges_from([(1, 2), (2, 3), (3, 2)])
# 提取弱连通分量
weak = list(nx.weakly_connected_components(G))
print("弱连通分量:", weak)
# 输出: 弱连通分量: [{1, 2, 3}]
# 提取强连通分量
strong = list(nx.strongly_connected_components(G))
print("强连通分量:", strong)
# 输出: 强连通分量: [{2, 3}, {1}]
````

### 📋 总结
| 概念 | 前提 | 核心规则 | 比喻 | 严苛程度 |
| :--- | :--- | :--- | :--- | :--- |
| **弱连通分量** | 有向图 | **忽略方向**，物理连通即可 | 无视单行道标志，能走通就行 | 宽松（容易满足） |
| **强连通分量** | 有向图 | **遵循方向**，必须互相可达 | 遵守交规，去得容易回得来才算 | 严苛（很难满足） |

**划重点：**
1. 强连通分量里的节点，一定也在同一个弱连通分量里。（强包含于弱）
2. 一个图可能整体是一个弱连通分量，但被拆成好几个碎片的强连通分量。
3. 现实意义：强连通分量往往代表一个**“利益闭环”**（比如网页互相引用、人群互相讨论），而弱连通只代表**“有牵连”**。

````python
# 整体网络指标
density = nx.density(G)
weak_components = list(nx.weakly_connected_components(G))
strong_components = list(nx.strongly_connected_components(G))
avg_clustering = nx.average_clustering(G.to_undirected())

print("网络密度：", round(density, 4))
print("弱连通分量数量：", len(weak_components))
print("强连通分量数量：", len(strong_components))
print("平均聚类系数：", round(avg_clustering, 4))

# 平均最短路径长度通常在最大弱连通子图上计算
largest_component_nodes = max(weak_components, key=len)
G_largest = G.subgraph(largest_component_nodes).to_undirected()

if nx.is_connected(G_largest):
    avg_path_length = nx.average_shortest_path_length(G_largest)
    print("最大弱连通子图平均最短路径长度：", round(avg_path_length, 4))
else:
    print("最大弱连通子图仍非连通，不能直接计算平均最短路径长度。")
````

Pyvis 的交互性非常强，它的参数主要分为三类：**物理引擎控制**、**元素交互属性**（节点/边）、以及**全局交互开关**。

### 🎛️ 表一：物理引擎与布局控制（最核心的交互）

### 🎛️ Pyvis 物理引擎参数详解
#### **barnesHut（Barnes-Hut 算法，默认求解器）**
| 参数 | 作用 | 调节建议 |
| :--- | :--- | :--- |
| **theta** | 精度与性能的平衡点（0~1） | 越小越精确，但计算越慢。默认 0.5 是个不错的平衡点。 |
| **gravitationalConstant** | 全局引力强度（负值表示引力） | -2000 是个常见值，值越大（负得越多），节点越往中心靠。 |
| **centralGravity** | 中心引力强度 | 0.3 是默认值，值越大，节点越向中心聚集。 |
| **springLength** | 弹簧自然长度（节点间理想距离） | 95 是默认值，值越大，节点间距越大。 |
| **springConstant** | 弹簧弹性系数 | 0.04 是默认值，值越大，边越“紧绷”。 |
| **damping** | 阻尼系数（防止震荡） | 0.09 是默认值，值越大，节点运动越“稳”。 |
| **avoidOverlap** | 避免节点重叠 | 0 表示不禁用，设为 1 可防止节点重叠。 |
| **maxVelocity** | 节点最大速度 | 50 是默认值，防止节点运动过快。 |
| **minVelocity** | 节点最小速度 | 0.75 是默认值，防止节点运动过慢。 |
| **solver** | 物理求解器 | `barnesHut`（默认）或 `forceAtlas2Based`。 |
| **timestep** | 时间步长 | 0.5 是默认值，值越小，运动越平滑。 |
#### **wind（风力效果）**
| 参数 | 作用 | 调节建议 |
| :--- | :--- | :--- |
| **x** | 水平方向风力 | 0 表示无风，正值向右吹，负值向左吹。 |
| **y** | 垂直方向风力 | 0 表示无风，正值向下吹，负值向上吹。 |



### 📋 调参口诀

> **“先调引力（gravitationalConstant），再调弹簧（springLength），最后加阻尼（damping）防抖。”**
> 
> 1. **引力**决定节点是否抱团（负值越大越抱团）。
> 2. **弹簧长度**决定节点间距（值越大越疏远）。
> 3. **阻尼**防止节点乱晃（值越大越稳定）。

## 第二部分：Plotly —— 精准悬停的“数据图”
Plotly 画网络图**没有物理引擎**，节点不能拖拽。但它的优势是：**鼠标悬停时，能极其精美地展示数据**，而且画极大规模图时性能更好。
画 Plotly 网络图的核心逻辑是：**把 NetworkX 的数据拆成“散点（节点）”和“折线（边）”画在一张图上。**

````python
import networkx as nx
import plotly.graph_objects as go

# 1. 创建图并算布局
G = nx.karate_club_graph()
pos = nx.spring_layout(G, seed=42, k=0.5) 

# --------------------------------------------------
# 第一步：画边（用折线 Scatter）
# --------------------------------------------------
edge_x = []
edge_y = []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None]) 
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=0.5, color='#888'),
    hoverinfo='none', 
    mode='lines')

# --------------------------------------------------
# 第二步：画节点（用散点 Scatter）
# --------------------------------------------------
node_x = []
node_y = []
node_text = []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_text.append(f'节点: {node}<br>度数: {G.degree(node)}')

node_trace = go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text', 
    text=[f"N{n}" for n in G.nodes()], 
    textposition="top center",
    hoverinfo='text',     
    hovertext=node_text,  
    marker=dict(
        showscale=True,   
        colorscale='YlGnBu',
        size=10,
        colorbar=dict(thickness=15, title='Node Connections', xanchor='left')
    ))

# 给节点按度数上色
node_adjacencies = []
for node, adjacencies in enumerate(G.adjacency()):
    node_adjacencies.append(len(adjacencies[1]))
node_trace.marker.color = node_adjacencies

# --------------------------------------------------
# 第三步：组合出图并输出为 HTML
# --------------------------------------------------
fig = go.Figure(data=[edge_trace, node_trace],
             layout=go.Layout(
                title=dict(text='Plotly 交互式网络图', font=dict(size=16)),
                showlegend=False,
                hovermode='closest', 
                margin=dict(b=20, l=5, r=5, t=40),
                xaxis=dict(showgrid=False, zeroline=False, showticklabels=False), 
                yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                )

# 🌟 关键修改：不再用 fig.show()，直接写成 HTML 文件
output_file = "plotly_network.html"
fig.write_html(output_file)

print(f"✅ 网络图已生成！请在当前文件夹找到 [{output_file}]，双击用浏览器打开查看。")
````


## 📋 画网络图的两种思路
| 工具 | 核心比喻 | 操作逻辑 | 特点 |
| :--- | :--- | :--- | :--- |
| **Pyvis** | 🚚 **"搬运工"** | NetworkX 建图 → 一键导入 Pyvis<br>`nt.from_nx(G)` 搞定！ | 物理引擎自动算坐标和动画，**省心省力** |
| **Plotly** | 🔧 **"拆解工"** | 1. NetworkX 算坐标 (`layout`)<br>2. 拆出边的 X,Y → 画折线<br>3. 拆出点的 X,Y → 画散点<br>4. 拼在一起出图 | 需要自己组装，但**控制力最强**，数据展示极佳 |
> **💡 共同归宿**：不管用哪种思路，交互图的最终交付方式都是 **`write_html()`**，用浏览器打开才能体验完整交互！

完整代码：

````python
import cv2
import networkx as nx
import numpy as np
from matplotlib import pyplot as plt
 
# detector parameters
block_size = 3
sobel_size = 3
k = 0.06
 
image = cv2.imread("img/08.jpg")
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray_img = np.float32(gray_img)
 
# detect the corners with appropriate values as input parameters
corners_img = cv2.cornerHarris(gray_img, block_size, sobel_size, k)
 
# set the number of keypoints to be selected
num_keypoints = 50
 
# find the top num_keypoints corners in the corners_img
corners = cv2.goodFeaturesToTrack(gray_img, num_keypoints, 0.01, 10)
corners = np.intp(corners)
 
# create an empty feature vector with n dimensions
feature_vector = np.zeros((num_keypoints, 2))
 
# iterate over the corners and construct the feature vector
for i, corner in enumerate(corners):
    x, y = corner.ravel()
    feature_vector[i] = [x, y]
    # print the extracted keypoints
    print("Keypoint %d: x=%d, y=%d" % (i+1, x, y))
 
# create an empty graph
G = nx.Graph()
 
# add nodes to the graph
for i, feature in enumerate(feature_vector):
    x, y = feature
    G.add_node(i, pos=(x, y))
 
# add edges to the graph (connecting each node to its two closest neighbors)
for i in range(len(feature_vector)):
    distances = np.sqrt(np.sum((feature_vector - feature_vector[i])**2, axis=1))
    closest_neighbors = np.argsort(distances)[1:3] # exclude the node itself
    for neighbor in closest_neighbors:
        G.add_edge(i, neighbor, weight=1)
 
# draw the graph
pos = nx.get_node_attributes(G, 'pos')
nx.draw_networkx(G, pos=pos, with_labels=False, node_size=50)
 
# show the image with the detected corners and feature vector
for r in range(image.shape[0]):
    for c in range(image.shape[1]):
        pix = corners_img[r, c]
        if pix > 0.05 * corners_img.max():
            cv2.circle(image, (c, r), 5, (0, 0, 255), 0)
 
# plot the feature vector on the image
for i in range(num_keypoints):
    x, y = feature_vector[i]
    cv2.circle(image, (int(x), int(y)), 10, (0, 255, 0), -1)
 
# convert the image back to RGB format
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
 
# display the image
plt.imshow(image)
plt.show()
````

## 有向图的构造：

````python
import cv2
import networkx as nx
import numpy as np
from matplotlib import pyplot as plt
 
# detector parameters
block_size = 3
sobel_size = 3
k = 0.06
 
image = cv2.imread("img/03.PNG")
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray_img = np.float32(gray_img)
 
# detect the corners with appropriate values as input parameters
corners_img = cv2.cornerHarris(gray_img, block_size, sobel_size, k)
 
# set the number of keypoints to be selected
num_keypoints = 50
 
# find the top num_keypoints corners in the corners_img
corners = cv2.goodFeaturesToTrack(gray_img, num_keypoints, 0.01, 10)
corners = np.intp(corners)
 
# create an empty feature vector with n dimensions
feature_vector = np.zeros((num_keypoints, 2))
 
# iterate over the corners and construct the feature vector
for i, corner in enumerate(corners):
    x, y = corner.ravel()
    feature_vector[i] = [x, y]
 
# create an empty directed graph
G = nx.DiGraph()
 
# add nodes to the graph
for i, feature in enumerate(feature_vector):
    x, y = feature
    G.add_node(i, pos=(x, y))
 
# add edges to the graph
for i in range(len(feature_vector)):
    distances = np.sqrt(np.sum((feature_vector[i] - feature_vector)**2, axis=1))
    closest_indices = np.argsort(distances)[1:3]
    for j in closest_indices:
        G.add_edge(i, j, weight=1)
 
# draw the graph
pos = nx.get_node_attributes(G, 'pos')
nx.draw_networkx(G, pos=pos, with_labels=False, node_size=50)
 
# show the image with the detected corners and feature vector
for r in range(image.shape[0]):
    for c in range(image.shape[1]):
        pix = corners_img[r, c]
        if pix > 0.05 * corners_img.max():
            cv2.circle(image, (c, r), 5, (0, 0, 255), 0)
 
# plot the feature vector on the image
for i in range(num_keypoints):
    x, y = feature_vector[i]
    cv2.circle(image, (int(x), int(y)), 10, (0, 255, 0), -1)
 
# convert the image back to RGB format
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
 
# display the image
plt.imshow(image)
plt.show()
````
